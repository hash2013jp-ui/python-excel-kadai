```python
import openpyxl
import pandas as pd
from openpyxl.styles import PatternFill

file_list = ['2022_年間売上表.xlsx','2023_年間売上表.xlsx']

df = pd.DataFrame()

for file_name in file_list:
  temp_df = pd.read_excel(file_name, sheet_name='Sheet1')
  df = pd.concat([df,temp_df],ignore_index=True)

group = df.groupby(['商品','売上年'])['金額（千円）'].sum()
group_df =group.reset_index()

writer = pd.ExcelWriter('売上集計表.xlsx', engine='openpyxl')
group_df.to_excel(writer,sheet_name='売上集計表',index=False)

sheet = writer.sheets['売上集計表']

for cell in sheet[1]:
    cell.fill = PatternFill(
        patternType='solid',
        fgColor='F2F2F2'
    )

writer.close()


```
