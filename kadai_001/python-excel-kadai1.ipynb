```python
import openpyxl
from datetime import datetime

workbook = openpyxl.Workbook()
ws = workbook.active

# 請求書様式を記入
ws["b2"] = "請求書"
ws["f4"] = "No."
ws["f5"] = "日付"
ws["g4"] = "0001"
ws["g5"] = datetime.now().strftime("%Y/%m/%d")

data = [
    ["商品名", "数量", "単価", "金額"],
    ["商品A", None, 10000, "=C11*D11"],
    ["商品B", None, 15000, "=C12*D12"],
    [None, None, None, "=SUM(E11:E12)"],
    [None],
    ["小計", None, None, "=E13"],
    ["消費税", None, None, "=E15*0.1"],
    ["合計", None, None, "=E15+E16"]
]

# B10セルからの請求書様式を記入
start_row = 10

for i, row in enumerate(data):
  current_row = start_row + i
  for j, value in enumerate(row):
    ws.cell(row=current_row, column=2 + j, value=value)

# 請求先データを記入

kokyaku_data = [
    "株式会社ＡＢＣ",
    "〒101-0022 東京都千代田区神田練塀町300",
    "TEL:03-1234-5678 FAX:03-1234-5678",
    "担当者名:鈴木一郎 様"
]

for i, row in enumerate(kokyaku_data):
  ws.cell(row=i+4, column=2, value=row)

#請求先の数量の記入
suryou_data = [
    ["株式会社ＢＢＣ", 1, 1],
    ["株式会社ＡＢＣ", 2, 1]
]

# リストから各要素（企業名、商品A数量、商品B数量）を展開してループ処理
seikyu_saki = ws["B4"].value

for kigyou_mei, price_a, price_b in suryou_data:
  if kigyou_mei == seikyu_saki:
    ws.cell(row=11, column=3, value=price_a)
    ws.cell(row=12, column=3, value=price_b)

# タイムスタンプ付きファイル名で保存
today = datetime.today()
filename = f"請求書_{today.strftime('%Y%m%d')}.xlsx"

workbook.save(filename)
```
